```python
# ============================================================
# 24. CORRECT P&L ENGINE
# ============================================================

def get_dataset_exit_pnl(row, horizon):
    """
    Use the dataset's actual CALL/PUT exit P&L.

    Direction:
        1  = CALL
       -1  = PUT
    """

    direction = row["predicted_direction"]

    if horizon == "15m":
        call_col = "15_call_Exit"
        put_col = "15_put_Exit"

    elif horizon == "30m":
        call_col = "30_call_Exit"
        put_col = "30_put_Exit"

    elif horizon == "60m":
        call_col = "60_call_Exit"
        put_col = "60_put_Exit"

    else:
        return np.nan

    if direction == 1:
        return row[call_col]

    elif direction == -1:
        return row[put_col]

    return np.nan


def calculate_correct_pnl(trades, horizon):
    """
    Calculate P&L using the actual dataset exit columns.
    """

    trades = trades.copy()

    if len(trades) == 0:
        return trades

    # 10m does not have official CALL/PUT exit columns
    if horizon == "10m":
        trades["pnl"] = np.nan
        trades["pnl_source"] = "NO_10M_EXIT_COLUMN"
        return trades

    trades["pnl"] = trades.apply(
        lambda row: get_dataset_exit_pnl(
            row,
            horizon
        ),
        axis=1
    )

    trades["pnl_source"] = "DATASET_EXIT"

    return trades
```

```python
# ============================================================
# CORRECTED PERFORMANCE
# ============================================================

def corrected_performance(trades):

    trades = trades.copy()

    # Check if the DataFrame is empty or if the 'pnl' column is missing
    # This handles cases like '10m' where an empty DataFrame without 'pnl' is passed.
    if len(trades) == 0 or "pnl" not in trades.columns:

        return {
            "trades": 0,
            "win_rate": np.nan,
            "avg_pnl": np.nan,
            "total_pnl": np.nan,
            "profit_factor": np.nan,
            "max_drawdown": np.nan
        }

    # Remove unavailable P&L
    trades = trades[
        trades["pnl"].notna()
    ].copy()

    # Re-check if trades became empty after removing NaN P&L
    if len(trades) == 0:
        return {
            "trades": 0,
            "win_rate": np.nan,
            "avg_pnl": np.nan,
            "total_pnl": np.nan,
            "profit_factor": np.nan,
            "max_drawdown": np.nan
        }

    wins = trades["pnl"] > 0
    losses = trades["pnl"] < 0

    win_rate = (
        wins.mean() * 100
    )

    avg_pnl = trades["pnl"].mean()

    total_pnl = trades["pnl"].sum()

    gross_profit = (
        trades.loc[wins, "pnl"].sum()
    )

    gross_loss = abs(
        trades.loc[losses, "pnl"].sum()
    )

    if gross_loss > 0:
        profit_factor = (
            gross_profit / gross_loss
        )
    elif gross_profit > 0:
        profit_factor = np.inf
    else:
        profit_factor = 0

    equity = trades["pnl"].cumsum()

    running_max = equity.cummax()

    drawdown = (
        equity - running_max
    )

    max_drawdown = drawdown.min()

    return {
        "trades": len(trades),
        "win_rate": win_rate,
        "avg_pnl": avg_pnl,
        "total_pnl": total_pnl,
        "profit_factor": profit_factor,
        "max_drawdown": max_drawdown
    }
```

```python
# ============================================================
# STEP 24 - REAL MODELING PIPELINE
# TARGET CONSTRUCTION
# ============================================================

import pandas as pd
import numpy as np

print("=" * 80)
print("REAL MODELING PIPELINE")
print("=" * 80)

# Work from the original dataframe so that the official
# *_call_Exit / *_put_Exit columns remain available.
model_df = df.copy()

model_df["datetime"] = pd.to_datetime(
    model_df["datetime"],
    errors="coerce"
)

model_df = (
    model_df
    .sort_values("datetime")
    .reset_index(drop=True)
)

# ------------------------------------------------------------
# Check required columns
# ------------------------------------------------------------

required_exit_columns = [
    "15_call_Exit",
    "15_put_Exit",
    "30_call_Exit",
    "30_put_Exit",
    "60_call_Exit",
    "60_put_Exit"
]

missing_exit_columns = [
    c for c in required_exit_columns
    if c not in model_df.columns
]

if missing_exit_columns:
    raise ValueError(
        f"Missing required exit columns: {missing_exit_columns}"
    )

print("Required exit columns found.")

# ------------------------------------------------------------
# Create targets from the actual dataset P&L columns
# ------------------------------------------------------------

# CALL = 1
# PUT  = -1
#
# At each horizon, choose the direction whose official
# dataset exit P&L is larger.
#
# If both are <= 0, label as 0 = NO TRADE.

def create_direction_target(
    data,
    call_column,
    put_column
):

    call_pnl = pd.to_numeric(
        data[call_column],
        errors="coerce"
    )

    put_pnl = pd.to_numeric(
        data[put_column],
        errors="coerce"
    )

    target = np.where(
        (call_pnl > 0) & (call_pnl >= put_pnl),
        1,
        np.where(
            (put_pnl > 0) & (put_pnl > call_pnl),
            -1,
            0
        )
    )

    return pd.Series(
        target,
        index=data.index
    )


model_df["target_15m"] = create_direction_target(
    model_df,
    "15_call_Exit",
    "15_put_Exit"
)

model_df["target_30m"] = create_direction_target(
    model_df,
    "30_call_Exit",
    "30_put_Exit"
)

model_df["target_60m"] = create_direction_target(
    model_df,
    "60_call_Exit",
    "60_put_Exit"
)

print()
print("=" * 80)
print("TARGET DISTRIBUTIONS")
print("=" * 80)

for horizon in ["15m", "30m", "60m"]:

    target_col = f"target_{horizon}"

    print()
    print(horizon)

    print(
        model_df[target_col]
        .value_counts()
        .sort_index()
    )

    print(
        model_df[target_col]
        .value_counts(normalize=True)
        .sort_index()
        .round(4)
    )
```

```python
# ============================================================
# MODEL FEATURES
# ============================================================

print("=" * 80)
print("BUILDING MODEL FEATURES")
print("=" * 80)

# Columns that must NEVER be used as model features
exclude_columns = set([
    "datetime",
    "date",
    "time",

    # Future/P&L information
    "15_call_Exit",
    "15_put_Exit",
    "30_call_Exit",
    "30_put_Exit",
    "60_call_Exit",
    "60_put_Exit",

    # Targets
    "target_15m",
    "target_30m",
    "target_60m"
])

# Also exclude any target/exit/pnl-like columns
# to prevent future information leakage.
for col in model_df.columns:

    col_lower = col.lower()

    if (
        "target" in col_lower
        or "exit" in col_lower
        or "pnl" in col_lower
        or "profit" in col_lower
        or "label" in col_lower
    ):
        exclude_columns.add(col)

candidate_features = [
    col
    for col in model_df.columns
    if col not in exclude_columns
]

# Keep numeric features only
numeric_features = [
    col
    for col in candidate_features
    if pd.api.types.is_numeric_dtype(
        model_df[col]
    )
]

print(
    "Candidate numeric features:",
    len(numeric_features)
)

print(
    "First 30 features:"
)

print(
    numeric_features[:30]
)
```

```python
# ============================================================
# CLEAN MODEL FEATURES
# ============================================================

X = model_df[
    numeric_features
].copy()

# Replace infinite values
X = X.replace(
    [np.inf, -np.inf],
    np.nan
)

# Remove columns that are completely empty
all_nan_columns = [
    col
    for col in X.columns
    if X[col].isna().all()
]

if all_nan_columns:

    print(
        "Removing all-NaN columns:",
        len(all_nan_columns)
    )

    X = X.drop(
        columns=all_nan_columns
    )

# Forward fill only using past information
X = X.ffill()

# Remaining NaNs occur mainly at the beginning
X = X.fillna(0)

print()
print("Final X shape:", X.shape)

print(
    "Remaining NaNs:",
    X.isna().sum().sum()
)

print(
    "Remaining infinite values:",
    np.isinf(X.to_numpy()).sum()
)
```

```python
# ============================================================
# TIME-BASED SPLIT
# ============================================================

split_time = model_df["datetime"].quantile(
    0.80
)

train_mask = (
    model_df["datetime"] <= split_time
)

test_mask = (
    model_df["datetime"] > split_time
)

X_train = X.loc[train_mask].copy()
X_test = X.loc[test_mask].copy()

print("=" * 80)
print("TIME SPLIT")
print("=" * 80)

print(
    "Train:",
    len(X_train),
    "rows"
)

print(
    "Test:",
    len(X_test),
    "rows"
)

print(
    "Train start:",
    model_df.loc[
        train_mask,
        "datetime"
    ].min()
)

print(
    "Train end:",
    model_df.loc[
        train_mask,
        "datetime"
    ].max()
)

print(
    "Test start:",
    model_df.loc[
        test_mask,
        "datetime"
    ].min()
)

print(
    "Test end:",
    model_df.loc[
        test_mask,
        "datetime"
    ].max()
)
```

```python
# ============================================================
# STEP 24A - TRAIN REAL MODELS
# ============================================================

import numpy as np
import pandas as pd

from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

# Try LightGBM
try:
    from lightgbm import LGBMClassifier
except ImportError:
    !pip -q install lightgbm
    from lightgbm import LGBMClassifier


print("=" * 80)
print("TRAINING REAL WALK-FORWARD MODELS")
print("=" * 80)

# ------------------------------------------------------------
# Make sure X follows model_df exactly
# ------------------------------------------------------------

X = X.replace(
    [np.inf, -np.inf],
    np.nan
).fillna(0)

X_train = X.loc[train_mask].copy()
X_test = X.loc[test_mask].copy()

print("Training rows:", len(X_train))
print("Testing rows :", len(X_test))
print("Features     :", X_train.shape[1])


models = {}
predictions = {}

horizons = ["15m", "30m", "60m"]


for horizon in horizons:

    target_col = f"target_{horizon}"

    y = model_df[target_col].astype(int)

    y_train = y.loc[train_mask]
    y_test = y.loc[test_mask]

    print()
    print("=" * 80)
    print(f"TRAINING {horizon} MODEL")
    print("=" * 80)

    print("Training target distribution:")
    print(y_train.value_counts().sort_index())

    model = LGBMClassifier(
        objective="multiclass",
        num_class=3,
        n_estimators=300,
        learning_rate=0.03,
        num_leaves=31,
        max_depth=-1,
        subsample=0.8,
        colsample_bytree=0.8,
        random_state=42,
        n_jobs=-1,
        verbosity=-1
    )

    model.fit(
        X_train,
        y_train
    )

    pred = model.predict(X_test)

    pred = pd.Series(
        pred,
        index=X_test.index,
        name=f"prediction_{horizon}"
    ).astype(int)

    models[horizon] = model
    predictions[horizon] = pred

    print()
    print("Test prediction distribution:")
    print(pred.value_counts().sort_index())

    print()
    print("Accuracy:")
    print(
        accuracy_score(
            y_test,
            pred
        )
    )

    print()
    print("Classification report:")

    print(
        classification_report(
            y_test,
            pred,
            labels=[-1, 0, 1],
            target_names=[
                "PUT",
                "NO_TRADE",
                "CALL"
            ],
            zero_division=0
        )
    )
```

```python
# ============================================================
# STEP 24C - CREATE REAL TRADES
# ============================================================

real_trades = {}


for horizon in horizons:

    prediction_column = (
        f"prediction_{horizon}"
    )

    trades = test_predictions[
        test_predictions[prediction_column].isin(
            [-1, 1]
        )
    ].copy()

    trades["predicted_direction"] = (
        trades[prediction_column]
    )

    trades = trades.drop(
        columns=[prediction_column]
    )

    real_trades[horizon] = trades

    print()
    print("=" * 80)
    print(f"{horizon} REAL TRADES")
    print("=" * 80)

    print(
        "Number of trades:",
        len(trades)
    )

    print(
        "CALL:",
        (trades["predicted_direction"] == 1).sum()
    )

    print(
        "PUT :",
        (trades["predicted_direction"] == -1).sum()
    )
```

```python
# ============================================================
# STEP 24D - OFFICIAL DATASET P&L
# ============================================================

exit_mapping = {

    "15m": {
        1: "15_call_Exit",
        -1: "15_put_Exit"
    },

    "30m": {
        1: "30_call_Exit",
        -1: "30_put_Exit"
    },

    "60m": {
        1: "60_call_Exit",
        -1: "60_put_Exit"
    }
}


for horizon in horizons:

    trades = real_trades[horizon].copy()

    call_column = exit_mapping[horizon][1]
    put_column = exit_mapping[horizon][-1]

    trades["pnl"] = np.where(
        trades["predicted_direction"] == 1,
        trades[call_column],
        trades[put_column]
    )

    trades["pnl"] = pd.to_numeric(
        trades["pnl"],
        errors="coerce"
    )

    real_trades[horizon] = trades


    print()
    print("=" * 80)
    print(f"{horizon} P&L CHECK")
    print("=" * 80)

    print(
        "Trades:",
        len(trades)
    )

    print(
        "Valid P&L:",
        trades["pnl"].notna().sum()
    )

    print(
        "Missing P&L:",
        trades["pnl"].isna().sum()
    )

    print(
        "Total P&L:",
        trades["pnl"].sum()
    )
```

```python
# ============================================================
# STEP 24E - PERFORMANCE ENGINE
# ============================================================

def calculate_final_performance(trades):

    trades = trades.copy()

    trades = trades[
        trades["pnl"].notna()
    ].copy()

    if len(trades) == 0:

        return {
            "trades": 0,
            "win_rate": 0,
            "avg_pnl": 0,
            "total_pnl": 0,
            "profit_factor": 0,
            "max_drawdown": 0,
            "max_drawdown_pct": 0,
            "trades_per_day": 0
        }

    trades = trades.sort_values(
        "datetime"
    )

    pnl = trades["pnl"]

    wins = pnl > 0
    losses = pnl < 0

    win_rate = (
        wins.mean() * 100
    )

    avg_pnl = pnl.mean()

    total_pnl = pnl.sum()

    gross_profit = pnl[
        pnl > 0
    ].sum()

    gross_loss = abs(
        pnl[
            pnl < 0
        ].sum()
    )

    if gross_loss > 0:
        profit_factor = (
            gross_profit /
            gross_loss
        )

    elif gross_profit > 0:
        profit_factor = np.inf

    else:
        profit_factor = 0

    equity = pnl.cumsum()

    running_max = equity.cummax()

    drawdown = (
        equity -
        running_max
    )

    max_drawdown = drawdown.min()

    # Percentage drawdown
    equity_base = (
        running_max.abs()
        .replace(0, np.nan)
    )

    drawdown_pct = (
        drawdown /
        equity_base *
        100
    )

    max_drawdown_pct = (
        drawdown_pct.min()
        if drawdown_pct.notna().any()
        else 0
    )

    number_of_days = (
        trades["datetime"]
        .dt.date
        .nunique()
    )

    trades_per_day = (
        len(trades) /
        number_of_days
        if number_of_days > 0
        else 0
    )

    return {
        "trades": len(trades),
        "win_rate": win_rate,
        "avg_pnl": avg_pnl,
        "total_pnl": total_pnl,
        "profit_factor": profit_factor,
        "max_drawdown": max_drawdown,
        "max_drawdown_pct": max_drawdown_pct,
        "trades_per_day": trades_per_day
    }
```

```python
# ============================================================
# STEP 24F - FINAL INTRADAY RESULTS
# ============================================================

intraday_results = {}


for horizon in horizons:

    result = calculate_final_performance(
        real_trades[horizon]
    )

    intraday_results[horizon] = result

    print()
    print("=" * 80)
    print(f"{horizon} FINAL BACKTEST")
    print("=" * 80)

    for key, value in result.items():

        if isinstance(value, float):

            if np.isinf(value):
                print(
                    f"{key}: INF"
                )
            else:
                print(
                    f"{key}: {value:.4f}"
                )

        else:

            print(
                f"{key}: {value}"
            )
```

```python
daily_open["spot_0930"] = daily_open["spot"]
daily_close["spot_1530"] = daily_close["spot"]

# Sort by date
daily_open = daily_open.sort_values("date").reset_index(drop=True)

# Next trading day's date and 09:30 price
daily_open["next_trade_date"] = daily_open["date"].shift(-1)
daily_open["next_day_0930"] = daily_open["spot_0930"].shift(-1)

# Merge next-day 09:30 information with today's 15:30
overnight_daily = daily_close.merge(
    daily_open[
        ["date", "next_trade_date", "next_day_0930"]
    ],
    on="date",
    how="left"
)

# Overnight return
overnight_daily["overnight_return"] = (
    overnight_daily["next_day_0930"]
    - overnight_daily["spot_1530"]
) / overnight_daily["spot_1530"]

print(overnight_daily.head(10))
```

```python
OVERNIGHT_THRESHOLD = 0.005   # 0.5%

overnight_daily["overnight_target"] = np.select(
    [
        overnight_daily["overnight_return"] >= OVERNIGHT_THRESHOLD,
        overnight_daily["overnight_return"] <= -OVERNIGHT_THRESHOLD
    ],
    [
        1,   # CALL
        -1   # PUT
    ],
    default=0
)

print("=" * 80)
print("OVERNIGHT TARGET DISTRIBUTION")
print("=" * 80)

print(
    overnight_daily["overnight_target"]
    .value_counts()
    .sort_index()
)

print("\nPercentages:")
print(
    overnight_daily["overnight_target"]
    .value_counts(normalize=True)
    .sort_index() * 100
)
```

```python
def check_acceptance(row):

    horizon = row["horizon"]

    if horizon == "Overnight":
        target_win = 70.0
        target_avg = 2000.0
        target_pf = 2.20
        target_mdd = 10.0
    else:
        target_win = 60.0
        target_avg = 1000.0
        target_pf = 1.75
        target_mdd = 12.0

    return pd.Series({
        "win_rate_pass": row["win_rate"] >= target_win,
        "avg_pnl_pass": row["avg_pnl"] >= target_avg,
        "profit_factor_pass": row["profit_factor"] >= target_pf,
        "mdd_pass": abs(row["max_drawdown_pct"]) < target_mdd
    })


acceptance_check = final_results.apply(
    check_acceptance,
    axis=1
)

final_acceptance = pd.concat(
    [
        final_results[["horizon"]],
        acceptance_check
    ],
    axis=1
)

final_acceptance["ALL_CRITERIA_PASS"] = (
    final_acceptance[
        [
            "win_rate_pass",
            "avg_pnl_pass",
            "profit_factor_pass",
            "mdd_pass"
        ]
    ].all(axis=1)
)

print("=" * 100)
print("ACCEPTANCE CRITERIA")
print("=" * 100)

display(final_acceptance)
```
