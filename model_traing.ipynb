```python
def triple_barrier_labels(
    data,
    horizon_bars,
    limit
):
    """
    Generate Triple Barrier labels.

    +1 = CALL
    -1 = PUT
     0 = NEUTRAL

    Uses only future spot prices for TARGET generation.
    Features themselves remain T0/T-1 only.
    """

    labels = np.zeros(len(data), dtype=np.int8)

    spots = data["spot"].to_numpy()
    dates = data["date"].to_numpy()
    times = data["time"].to_numpy()

    for i in range(len(data)):

        entry_price = spots[i]

        if not np.isfinite(entry_price) or entry_price <= 0:
            continue

        current_date = dates[i]

        upper = entry_price * (1 + limit)
        lower = entry_price * (1 - limit)

        # Maximum allowed future index
        end_index = min(
            i + horizon_bars,
            len(data) - 1
        )

        # Search forward
        for j in range(i + 1, end_index + 1):

            # Never cross trading day
            if dates[j] != current_date:
                break

            future_price = spots[j]

            if not np.isfinite(future_price):
                continue

            # Upper barrier
            if future_price >= upper:
                labels[i] = 1
                break

            # Lower barrier
            if future_price <= lower:
                labels[i] = -1
                break

            # Stop at market close
            if times[j].strftime("%H:%M") >= "15:25":
                break

    return labels
```

```python
walk_forward_folds = []

for i in range(4, len(months)):

    train_months = months[:i]
    test_month = months[i]

    train_df = final_df[
        final_df["month"].isin(train_months)
    ]

    test_df = final_df[
        final_df["month"] == test_month
    ]

    walk_forward_folds.append(
        {
            "train_end": train_months[-1],
            "test_month": test_month,
            "train_df": train_df,
            "test_df": test_df
        }
    )

print(
    "Number of folds:",
    len(walk_forward_folds)
)
```

```python
def apply_embargo(train_df, test_df, minutes=60):

    test_start = test_df["datetime"].min()

    cutoff = (
        test_start -
        pd.Timedelta(minutes=minutes)
    )

    return train_df[
        train_df["datetime"] < cutoff
    ].copy()
```

```python
from lightgbm import LGBMClassifier
```

```python
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
```

```python
stage1_model = Pipeline([
    (
        "imputer",
        SimpleImputer(strategy="median")
    ),
    (
        "model",
        LGBMClassifier(
            n_estimators=300,
            learning_rate=0.03,
            num_leaves=31,
            max_depth=-1,
            subsample=0.8,
            colsample_bytree=0.8,
            class_weight="balanced",
            random_state=42,
            n_jobs=-1,
            verbosity=-1
        )
    )
])
```

```python
fold = walk_forward_folds[0]

train_df = fold["train_df"]
test_df = fold["test_df"]

train_df = apply_embargo(
    train_df,
    test_df,
    minutes=60
)

# Filter selected_features to only include columns actually present in train_df
actual_selected_features_train = [
    col for col in selected_features if col in train_df.columns
]
X_train = train_df[actual_selected_features_train]
y_train = train_df["stage1_target"]

# Filter selected_features for test_df as well
actual_selected_features_test = [
    col for col in selected_features if col in test_df.columns
]
X_test = test_df[actual_selected_features_test]
y_test = test_df["stage1_target"]

print("Train:", X_train.shape)
print("Test:", X_test.shape)

print("\nTrain target:")
print(y_train.value_counts())

print("\nTest target:")
print(y_test.value_counts())
```

```python
def calculate_trade_pnl(
    predicted_direction,
    target_label,
    entry_price,
    limit=0.01
):

    if predicted_direction == 0:
        return 0.0, 0.0

    # If upper barrier was hit
    if target_label == 1:
        realized_return = limit

    # If lower barrier was hit
    elif target_label == -1:
        realized_return = -limit

    # If no barrier was hit
    else:
        realized_return = 0.0

    if predicted_direction == 1:
        pnl = (
            realized_return * 5000
        ) - 200

    else:
        pnl = (
            (-realized_return) * 5000
        ) - 200

    return realized_return, pnl
```

```python
def calculate_performance(trades):

    if len(trades) == 0:
        return {
            "trades": 0,
            "win_rate": 0,
            "avg_pnl": 0,
            "total_pnl": 0,
            "profit_factor": 0,
            "max_drawdown": 0
        }

    trades = trades.copy()

    # Winning trades
    trades["win"] = trades["pnl"] > 0

    # Win rate
    win_rate = trades["win"].mean() * 100

    # Average P&L
    avg_pnl = trades["pnl"].mean()

    # Total P&L
    total_pnl = trades["pnl"].sum()

    # Gross profit
    gross_profit = trades.loc[
        trades["pnl"] > 0,
        "pnl"
    ].sum()

    # Gross loss
    gross_loss = abs(
        trades.loc[
            trades["pnl"] < 0,
            "pnl"
        ].sum()
    )

    # Profit factor
    profit_factor = (
        gross_profit / gross_loss
        if gross_loss > 0
        else np.inf
    )

    # Equity curve
    trades["equity"] = trades["pnl"].cumsum()

    # Running maximum
    running_max = trades["equity"].cummax()

    # Drawdown
    drawdown = trades["equity"] - running_max

    max_drawdown = drawdown.min()

    return {
        "trades": len(trades),
        "win_rate": win_rate,
        "avg_pnl": avg_pnl,
        "total_pnl": total_pnl,
        "profit_factor": profit_factor,
        "max_drawdown": max_drawdown
    }
```

```python
# ============================================================
# 8. WALK-FORWARD EMBARGO
# ============================================================

def apply_embargo(
    train_data,
    test_data,
    minutes=60
):

    test_start = (
        test_data["datetime"].min()
    )

    cutoff = (
        test_start
        -
        pd.Timedelta(
            minutes=minutes
        )
    )

    clean_train = train_data[
        train_data["datetime"] < cutoff
    ].copy()

    return clean_train


print("Embargo function ready.")
```

```python
# ============================================================
# 9. TRIPLE-BARRIER EXIT ENGINE
# ============================================================

def get_barrier_exit(
    data,
    index,
    horizon_bars,
    limit
):

    entry_row = data.iloc[index]

    entry_price = entry_row["spot"]
    entry_date = entry_row["date"]

    if (
        pd.isna(entry_price)
        or
        entry_price <= 0
    ):

        return {
            "barrier_label": 0,
            "exit_price": np.nan,
            "exit_datetime": pd.NaT,
            "exit_index": None,
            "return": np.nan
        }

    upper_price = (
        entry_price *
        (1 + limit)
    )

    lower_price = (
        entry_price *
        (1 - limit)
    )

    last_index = min(
        index + horizon_bars,
        len(data) - 1
    )

    vertical_index = None

    for j in range(
        index + 1,
        last_index + 1
    ):

        future_row = data.iloc[j]

        if (
            future_row["date"]
            != entry_date
        ):
            break

        future_price = (
            future_row["spot"]
        )

        if pd.isna(future_price):
            continue

        vertical_index = j

        if future_price >= upper_price:

            realized_return = (
                upper_price - entry_price
            ) / entry_price

            return {
                "barrier_label": 1,
                "exit_price": upper_price,
                "exit_datetime":
                    future_row["datetime"],
                "exit_index": j,
                "return": realized_return
            }

        if future_price <= lower_price:

            realized_return = (
                lower_price - entry_price
            ) / entry_price

            return {
                "barrier_label": -1,
                "exit_price": lower_price,
                "exit_datetime":
                    future_row["datetime"],
                "exit_index": j,
                "return": realized_return
            }

        if str(
            future_row["time"]
        ) >= "15:25:00":

            break

    if vertical_index is not None:

        exit_row = data.iloc[
            vertical_index
        ]

        exit_price = (
            exit_row["spot"]
        )

        realized_return = (
            exit_price - entry_price
        ) / entry_price

        return {
            "barrier_label": 0,
            "exit_price": exit_price,
            "exit_datetime":
                exit_row["datetime"],
            "exit_index":
                vertical_index,
            "return":
                realized_return
        }

    return {
        "barrier_label": 0,
        "exit_price": np.nan,
        "exit_datetime": pd.NaT,
        "exit_index": None,
        "return": np.nan
    }


print("Triple-barrier engine ready.")
```

```python
# ============================================================
# 10. P&L ENGINE
# ============================================================

def calculate_trade_pnl(
    direction,
    realized_return
):

    if direction == 0:
        return 0.0

    if pd.isna(realized_return):
        return np.nan

    NOTIONAL = 5000
    TRANSACTION_COST = 200

    if direction == 1:

        pnl = (
            realized_return *
            NOTIONAL
        ) - TRANSACTION_COST

    elif direction == -1:

        pnl = (
            -realized_return *
            NOTIONAL
        ) - TRANSACTION_COST

    else:

        pnl = 0.0

    return pnl


print("P&L engine ready.")
```

```python
# ============================================================
# 11. PERFORMANCE METRICS
# ============================================================

def calculate_performance(trades):

    if (
        trades is None
        or
        len(trades) == 0
    ):

        return {
            "trades": 0,
            "trades_per_day": 0,
            "win_rate": 0,
            "avg_pnl": 0,
            "total_pnl": 0,
            "profit_factor": 0,
            "max_drawdown": 0,
            "max_drawdown_pct": 0
        }

    trades = trades.copy()

    trades = (
        trades
        .sort_values(
            "entry_datetime"
        )
        .reset_index(drop=True)
    )

    trades["win"] = (
        trades["pnl"] > 0
    )

    win_rate = (
        trades["win"].mean() * 100
    )

    avg_pnl = (
        trades["pnl"].mean()
    )

    total_pnl = (
        trades["pnl"].sum()
    )

    gross_profit = (
        trades.loc[
            trades["pnl"] > 0,
            "pnl"
        ].sum()
    )

    gross_loss = abs(
        trades.loc[
            trades["pnl"] < 0,
            "pnl"
        ].sum()
    )

    if gross_loss > 0:

        profit_factor = (
            gross_profit /
            gross_loss
        )

    else:

        profit_factor = np.inf

    trades["equity"] = (
        trades["pnl"].cumsum()
    )

    running_max = (
        trades["equity"].cummax()
    )

    drawdown = (
        trades["equity"]
        -
        running_max
    )

    max_drawdown = (
        drawdown.min()
    )

    INITIAL_CAPITAL = 100000

    equity = (
        INITIAL_CAPITAL
        +
        trades["pnl"].cumsum()
    )

    equity_peak = (
        equity.cummax()
    )

    drawdown_pct = (
        (equity - equity_peak)
        /
        equity_peak
        * 100
    )

    max_drawdown_pct = (
        drawdown_pct.min()
    )

    trading_days = (
        trades[
            "entry_datetime"
        ]
        .dt.date
        .nunique()
    )

    if trading_days > 0:

        trades_per_day = (
            len(trades)
            /
            trading_days
        )

    else:

        trades_per_day = 0

    return {

        "trades":
            len(trades),

        "trades_per_day":
            trades_per_day,

        "win_rate":
            win_rate,

        "avg_pnl":
            avg_pnl,

        "total_pnl":
            total_pnl,

        "profit_factor":
            profit_factor,

        "max_drawdown":
            max_drawdown,

        "max_drawdown_pct":
            max_drawdown_pct
    }


print("Performance calculator ready.")
```

```python
# ============================================================
# 12. WALK-FORWARD BACKTEST ENGINE
# ============================================================

def run_walk_forward(
    data,
    feature_columns,
    target_column,
    horizon_bars,
    limit,
    config_name
):

    print()
    print("=" * 80)
    print(
        f"RUNNING {config_name.upper()} WALK-FORWARD BACKTEST"
    )
    print("=" * 80)

    working = data.copy()

    working["month"] = (
        working["datetime"]
        .dt.to_period("M")
    )

    months = sorted(
        working["month"]
        .dropna()
        .unique()
    )

    all_trades = []
    fold_results = []

    stage1_col = (
        f"{config_name}_stage1"
    )

    stage2_col = (
        f"{config_name}_stage2"
    )

    for fold_number in range(
        4,
        len(months)
    ):

        train_months = months[
            :fold_number
        ]

        test_month = months[
            fold_number
        ]

        train_data = (
            working[
                working["month"]
                .isin(train_months)
            ]
            .copy()
        )

        test_data = (
            working[
                working["month"]
                == test_month
            ]
            .copy()
        )

        if len(test_data) == 0:
            continue

        train_data = apply_embargo(
            train_data,
            test_data,
            minutes=60
        )

        if len(train_data) == 0:
            continue

        train_stage1 = (
            train_data
            .dropna(
                subset=[stage1_col]
            )
        )

        if (
            train_stage1[stage1_col]
            .nunique()
            < 2
        ):
            continue

        stage1_model = (
            create_stage1_model()
        )

        X_train = (
            train_stage1[
                feature_columns
            ]
        )

        y_train = (
            train_stage1[
                stage1_col
            ].astype(int)
        )

        X_test = (
            test_data[
                feature_columns
            ]
        )

        stage1_model.fit(
            X_train,
            y_train
        )

        train_probability = (
            stage1_model
            .predict_proba(
                X_train
            )[:, 1]
        )

        event_threshold = (
            np.quantile(
                train_probability,
                0.98
            )
        )

        test_probability = (
            stage1_model
            .predict_proba(
                X_test
            )[:, 1]
        )

        event_mask = (
            test_probability
            >=
            event_threshold
        )

        candidate_positions = (
            np.where(event_mask)[0]
        )

        stage2_train = (
            train_data[
                train_data[
                    target_column
                ].isin([-1, 1])
            ]
            .dropna(
                subset=[stage2_col]
            )
        )

        if (
            len(stage2_train) == 0
            or
            stage2_train[
                stage2_col
            ].nunique()
            < 2
        ):
            continue

        stage2_model = (
            create_stage2_model()
        )

        stage2_model.fit(

            stage2_train[
                feature_columns
            ],

            stage2_train[
                stage2_col
            ].astype(int)
        )

        if len(candidate_positions) == 0:

            print(
                f"Fold {fold_number:02d} | "
                f"{test_month} | "
                f"No candidates"
            )

            continue

        candidate_data = (
            test_data
            .iloc[
                candidate_positions
            ]
            .copy()
        )

        stage2_probability = (
            stage2_model
            .predict_proba(
                candidate_data[
                    feature_columns
                ]
            )[:, 1]
        )

        fold_trade_count = 0

        for candidate_number, original_position in enumerate(
            candidate_positions
        ):

            row = (
                test_data
                .iloc[
                    original_position
                ]
            )

            matching = (
                working.index[
                    working["datetime"]
                    ==
                    row["datetime"]
                ]
                .tolist()
            )

            if len(matching) == 0:
                continue

            global_index = (
                matching[0]
            )

            probability = (
                stage2_probability[
                    candidate_number
                ]
            )

            if probability >= 0.5:

                direction = 1

            else:

                direction = -1

            exit_info = (
                get_barrier_exit(

                    working,

                    global_index,

                    horizon_bars,

                    limit
                )
            )

            realized_return = (
                exit_info["return"]
            )

            if pd.isna(
                realized_return
            ):
                continue

            pnl = (
                calculate_trade_pnl(

                    direction,

                    realized_return
                )
            )

            trade = {

                "config":
                    config_name,

                "fold":
                    fold_number,

                "test_month":
                    str(test_month),

                "entry_datetime":
                    row["datetime"],

                "entry_price":
                    row["spot"],

                "exit_datetime":
                    exit_info[
                        "exit_datetime"
                    ],

                "exit_price":
                    exit_info[
                        "exit_price"
                    ],

                "realized_return":
                    realized_return,

                "actual_barrier":
                    exit_info[
                        "barrier_label"
                    ],

                "predicted_direction":
                    direction,

                "stage1_probability":
                    test_probability[
                        original_position
                    ],

                "stage1_threshold":
                    event_threshold,

                "stage2_probability":
                    probability,

                "pnl":
                    pnl
            }

            all_trades.append(
                trade
            )

            fold_trade_count += 1

        print(
            f"Fold {fold_number:02d} | "
            f"Test={test_month} | "
            f"Candidates={len(candidate_positions)} | "
            f"Trades={fold_trade_count}"
        )

        fold_results.append({

            "fold":
                fold_number,

            "test_month":
                str(test_month),

            "train_rows":
                len(train_data),

            "test_rows":
                len(test_data),

            "candidates":
                len(candidate_positions),

            "trades":
                fold_trade_count,

            "threshold":
                event_threshold
        })

    if len(all_trades) > 0:

        trades_df = pd.DataFrame(
            all_trades
        )

        trades_df = (
            trades_df
            .sort_values(
                "entry_datetime"
            )
            .reset_index(drop=True)
        )

    else:

        trades_df = pd.DataFrame()

    metrics = (
        calculate_performance(
            trades_df
        )
    )

    metrics["config"] = config_name

    metrics["horizon_minutes"] = (
        horizon_bars * 5
    )

    metrics["barrier_limit"] = limit

    return (
        trades_df,
        pd.DataFrame(fold_results),
        metrics
    )


print("Walk-forward engine ready.")
```

```python
# ============================================================
# 15. ACCEPTANCE CRITERIA
# ============================================================

def check_acceptance(row):

    win_pass = (
        row["win_rate"] >= 60
    )

    pnl_pass = (
        row["avg_pnl"] >= 1000
    )

    dd_pass = (
        abs(
            row["max_drawdown_pct"]
        ) < 12
    )

    pf_pass = (
        row["profit_factor"] >= 1.75
    )

    return pd.Series({

        "WinRate_PASS":
            win_pass,

        "AvgPnL_PASS":
            pnl_pass,

        "MaxDD_PASS":
            dd_pass,

        "ProfitFactor_PASS":
            pf_pass,

        "ALL_PASS":
            (
                win_pass
                and
                pnl_pass
                and
                dd_pass
                and
                pf_pass
            )
    })


acceptance_df = (
    results_df
    .apply(
        check_acceptance,
        axis=1
    )
)

acceptance_report = pd.concat(
    [
        results_df,
        acceptance_df
    ],
    axis=1
)

display(
    acceptance_report.round(3)
)
```

```python
# ============================================================
# 16. CALL VS PUT ANALYSIS
# ============================================================

for config_name, trades in (
    all_trades_results.items()
):

    print()
    print("=" * 70)
    print(
        f"{config_name} CALL VS PUT"
    )
    print("=" * 70)

    if len(trades) == 0:

        print("No trades.")

        continue

    temp = trades.copy()

    temp["direction"] = (
        temp[
            "predicted_direction"
        ]
        .map({
            1: "CALL",
            -1: "PUT"
        })
    )

    summary = (
        temp
        .groupby("direction")
        .agg(

            trades=(
                "pnl",
                "count"
            ),

            win_rate=(
                "pnl",
                lambda x:
                (x > 0).mean() * 100
            ),

            avg_pnl=(
                "pnl",
                "mean"
            ),

            total_pnl=(
                "pnl",
                "sum"
            )
        )
        .reset_index()
    )

    display(
        summary.round(3)
    )
```

```python
# ============================================================
# 17. MONTHLY PERFORMANCE
# ============================================================

monthly_results = {}

for config_name, trades in (
    all_trades_results.items()
):

    if len(trades) == 0:
        continue

    temp = trades.copy()

    temp["month"] = (
        temp["entry_datetime"]
        .dt.to_period("M")
        .astype(str)
    )

    monthly = (
        temp
        .groupby("month")
        .agg(

            trades=(
                "pnl",
                "count"
            ),

            total_pnl=(
                "pnl",
                "sum"
            ),

            avg_pnl=(
                "pnl",
                "mean"
            ),

            win_rate=(
                "pnl",
                lambda x:
                (x > 0).mean() * 100
            )
        )
        .reset_index()
    )

    monthly_results[
        config_name
    ] = monthly

    print()
    print(
        f"{config_name} MONTHLY RESULTS"
    )

    display(
        monthly.round(3)
    )
```

```python
# ============================================================
# 18. EQUITY CURVES
# ============================================================

for config_name, trades in (
    all_trades_results.items()
):

    if len(trades) == 0:
        continue

    trades = (
        trades
        .sort_values(
            "entry_datetime"
        )
    )

    equity = (
        trades["pnl"]
        .cumsum()
    )

    plt.figure(
        figsize=(14, 5)
    )

    plt.plot(
        trades["entry_datetime"],
        equity
    )

    plt.axhline(
        0,
        linestyle="--"
    )

    plt.title(
        f"{config_name} Equity Curve"
    )

    plt.xlabel(
        "Date"
    )

    plt.ylabel(
        "Cumulative P&L"
    )

    plt.xticks(
        rotation=45
    )

    plt.tight_layout()

    plt.show()
```
