# Part 1 — Basic NLP with NLTK

## 2. Import libraries

````python
import nltk

from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk import pos_tag, ne_chunk
````

# 3. Raw text

Start with a sentence that contains several NLP concepts:

````python
text = """
Ali is studying Artificial Intelligence at UET Peshawar.
He loves Natural Language Processing because it allows computers
to understand human language.
"""

print(text)
````

This is how the computer initially receives the text: simply as a string.

---

# 4. Sentence Tokenization

````python
sentences = sent_tokenize(text)

print(sentences)
````

**Sentence tokenization divides a document into individual sentences.**

---

# 5. Word Tokenization

````python
words = word_tokenize(text)

print(words)
````

**Word tokenization divides text into individual tokens.**

---

# 6. Lowercasing

````python
words_lower = [word.lower() for word in words]

print(words_lower)
````

For example, `Artificial` becomes `artificial`, `Intelligence` becomes `intelligence`. `Apple`, `apple`, and `APPLE` may represent the same word depending on the task.

---

# 7. Remove punctuation

````python
import string

words_clean = [
    word for word in words_lower
    if word not in string.punctuation
]

print(words_clean)
````

---

# 8. Stopword Removal

First see the stopwords:

````python
stop_words = set(stopwords.words("english"))

print(list(stop_words)[:30])
````

Now remove them:

````python
filtered_words = [
    word for word in words_clean
    if word not in stop_words
]

print(filtered_words)
````

For example, "is", "the", "to", "and" may be removed.

**Stopwords are commonly occurring words that may contribute little useful information for some NLP tasks.**

Important: we don't always remove stopwords. Modern Transformer models generally use the full tokenized input rather than manually deleting stopwords.

# Part 2 — Stemming

## 9. Porter Stemmer

````python
stemmer = PorterStemmer()

words = [
    "play",
    "playing",
    "played",
    "plays",
    "studies",
    "studying",
    "easily"
]

for word in words:
    print(word, "→", stemmer.stem(word))
````

**Stemming mechanically reduces words to a stem, and the resulting stem may not be a valid English word.**

---

# Part 3 — Lemmatization

## 10. WordNet Lemmatizer

````python
lemmatizer = WordNetLemmatizer()

words = [
    "playing",
    "played",
    "plays",
    "studies",
    "studying",
    "mice"
]

for word in words:
    print(word, "→", lemmatizer.lemmatize(word))
````

For better results, give the appropriate POS:

````python
print(lemmatizer.lemmatize("studies", pos="v"))
print(lemmatizer.lemmatize("playing", pos="v"))
print(lemmatizer.lemmatize("mice", pos="n"))
````

Compare: Stemming gives `studies → studi`, while Lemmatization gives `studies → study`.

---

# Part 4 — POS Tagging

## 11. Part-of-Speech Tagging

````python
sentence = "Ali is studying Artificial Intelligence."

tokens = word_tokenize(sentence)

pos_tags = pos_tag(tokens)

print(pos_tags)
````

Common tags: `NN` → noun, `VB` → verb, `JJ` → adjective, `RB` → adverb, `PRP` → pronoun. The exact tags depend on the tagger and context.

---

# Part 5 — Named Entity Recognition

## 12. NER with NLTK

````python
sentence = """
Ali is studying at UET Peshawar and wants to work at Google.
"""

tokens = word_tokenize(sentence)

pos_tags = pos_tag(tokens)

tree = ne_chunk(pos_tags)

print(tree)
````

NLTK may identify entities such as Ali → PERSON, UET Peshawar → ORGANIZATION, Google → ORGANIZATION.

**Named Entity Recognition identifies important entities such as people, organizations, locations, dates, etc.**

# Part 8 — Tokenizer

This is **very important** before teaching Transformer architecture.

````python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained(
    "bert-base-uncased"
)
````

````python
text = "I love Natural Language Processing."

tokens = tokenizer.tokenize(text)

print(tokens)
````

**A Transformer does not directly receive raw English. It receives tokens represented as numerical IDs.**

---

# 10. Convert tokens to IDs

````python
token_ids = tokenizer.convert_tokens_to_ids(tokens)

print(token_ids)
````

The exact IDs depend on the tokenizer/vocabulary.

```text
Text
 ↓
Tokens
 ↓
Token IDs
 ↓
Transformer
```

---

# 11. Let the tokenizer do everything

````python
inputs = tokenizer(
    "I love Natural Language Processing.",
    return_tensors="pt"
)

print(inputs)
````

You will see fields such as `input_ids` and `attention_mask`.

- **input_ids**: The numerical representation of tokens.
- **attention_mask**: Tells the model which positions contain actual input tokens versus padding.

---

# Part 9 — Understand Special Tokens

````python
text = "I love NLP."

encoded = tokenizer(
    text,
    return_tensors="pt"
)

print(encoded["input_ids"])
````

Decode it:

````python
print(
    tokenizer.convert_ids_to_tokens(
        encoded["input_ids"][0]
    )
)
````

For BERT, you'll typically see special tokens `[CLS]`, `i`, `love`, `nlp`, `.`, `[SEP]`. `[CLS]` is a special classification token; `[SEP]` separates/end-marks sequences.

---

# Part 10 — WordPiece Tokenization

````python
text = "unbelievable"

print(tokenizer.tokenize(text))
````

````python
words = [
    "playing",
    "unbelievable",
    "internationalization",
    "happiness"
]

for word in words:
    print(word, "→", tokenizer.tokenize(word))
````

**Transformers often use subword tokenization, so a word doesn't necessarily correspond to one token.** This is an important difference from the simple `word_tokenize()` students learned with NLTK.

---

# Part 11 — Attention Mask

````python
texts = [
    "I love NLP.",
    "I love machine learning."
]

encoded = tokenizer(
    texts,
    padding=True,
    return_tensors="pt"
)

print("Input IDs:")
print(encoded["input_ids"])

print("\nAttention Mask:")
print(encoded["attention_mask"])
````

`1` means real token, `0` means padding.

# Part 12 — Transformer Classification

````python
from transformers import AutoTokenizer, AutoModelForSequenceClassification

model_name = "distilbert-base-uncased-finetuned-sst-2-english"

tokenizer = AutoTokenizer.from_pretrained(model_name)

model = AutoModelForSequenceClassification.from_pretrained(
    model_name
)
````

````python
text = "I really enjoyed this course."

inputs = tokenizer(
    text,
    return_tensors="pt"
)

print(inputs)
````

````python
outputs = model(**inputs)

print(outputs.logits)
````

---

# Part 13 — Convert Logits to Probabilities

````python
import torch

probabilities = torch.softmax(
    outputs.logits,
    dim=-1
)

print(probabilities)
````

````python
predicted_class = torch.argmax(
    probabilities,
    dim=-1
)

print(predicted_class)
````

````python
print(model.config.id2label)
````

# Part 16 — Bag of Words (Before Embeddings)

Before jumping to embeddings, show students the "naive" numerical representation.

````python
from sklearn.feature_extraction.text import CountVectorizer

corpus = [
    "I love NLP",
    "I love machine learning",
    "NLP is fun"
]

vectorizer = CountVectorizer()
X = vectorizer.fit_transform(corpus)

print(vectorizer.get_feature_names_out())
print(X.toarray())
````

Explain:

> **Bag of Words counts word occurrences but ignores order and meaning.**

---

# Part 17 — TF-IDF

````python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer()
X_tfidf = tfidf.fit_transform(corpus)

print(tfidf.get_feature_names_out())
print(X_tfidf.toarray())
````

Explain:

> **TF-IDF downweights very common words and upweights words that are distinctive to a document.**

This is the classical bridge between "tokens" and "numbers" that pre-dates Transformers.

---

# Part 18 — Word Embeddings (Word2Vec)

````python
!pip install -q gensim
````

````python
from gensim.models import Word2Vec

sentences = [
    ["ali", "loves", "nlp"],
    ["ali", "studies", "ai"],
    ["nlp", "is", "part", "of", "ai"]
]

model = Word2Vec(sentences, vector_size=50, window=3, min_count=1)

print(model.wv["nlp"])
````

Explain:

> **Unlike Bag of Words, embeddings place words in a continuous vector space where similar words end up close together.**

Show similarity:

````python
print(model.wv.most_similar("nlp"))
````

---

# Part 19 — Cosine Similarity Between Words

````python
import numpy as np

def cosine_similarity(v1, v2):
    return np.dot(v1, v2) / (np.linalg.norm(v1) * np.linalg.norm(v2))

vec1 = model.wv["nlp"]
vec2 = model.wv["ai"]

print(cosine_similarity(vec1, vec2))
````

Explain:

> **Cosine similarity measures how close two vectors point in the same direction — the standard way to compare embeddings.**

# Part 20 — Sentence Embeddings with Transformers

Now connect embeddings to Hugging Face.

````python
!pip install -q sentence-transformers
````

````python
from sentence_transformers import SentenceTransformer, util

model = SentenceTransformer("all-MiniLM-L6-v2")

sentence1 = "I love studying Natural Language Processing."
sentence2 = "NLP is my favorite subject."
sentence3 = "The weather is nice today."

embeddings = model.encode([sentence1, sentence2, sentence3])

print(util.cos_sim(embeddings[0], embeddings[1]))
print(util.cos_sim(embeddings[0], embeddings[2]))
````

# Part 30 — Fine-Tuning, Conceptually

Students don't need to run full fine-tuning in class, but they should understand the idea before moving on.

```text
Pretrained model (knows language in general)
          +
Small labeled dataset (your specific task)
          ↓
   Fine-tuning (few epochs, small learning rate)
          ↓
   Task-specific model
```

````python
from transformers import TrainingArguments

training_args = TrainingArguments(
    output_dir="./results",
    num_train_epochs=3,
    per_device_train_batch_size=8,
    learning_rate=2e-5
)

print(training_args)
````

Explain:

> **Fine-tuning adjusts the pretrained weights slightly instead of learning from scratch — this is why it works with much less data than training a model from zero.**

# Part 31 — Updated Comparison Table

| Concept | Classical NLP (NLTK) | Modern NLP (Transformers) |
|---|---|---|
| Text → numbers | Bag of Words / TF-IDF | Token IDs → learned embeddings |
| Word meaning | Not captured | Captured via embeddings (Word2Vec, contextual embeddings) |
| Context sensitivity | None (word counts are static) | Full (self-attention, contextual) |
| Sequence handling | N/A | RNN/LSTM (sequential) → Transformer (parallel, attention-based) |
| Similarity | Cosine similarity on TF-IDF vectors | Cosine similarity on dense embeddings |
| New tasks | Requires new feature engineering | Zero-shot classification, prompting |
| Long documents | Struggles (sparse, no order) | Handles well (attention over full context) |
