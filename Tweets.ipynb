```python
def remove_urls(text):
    return re.sub(r"https?://\S+|www\.\S+", "", text)
```

```python
def handle_hashtags(text):
    return re.sub(r"#(\w+)", r"\1", text)
```

```python
def lowercase(text):
    return text.lower()
```

```python
def remove_punctuation(text):
    return text.translate(str.maketrans("", "", string.punctuation))
```

```python
def remove_emojis(text):
    emoji_pattern = re.compile(
        "["
        "\U0001F600-\U0001F64F"
        "\U0001F300-\U0001F5FF"
        "\U0001F680-\U0001F6FF"
        "\U0001F1E0-\U0001F1FF"
        "\U00002702-\U000027B0"
        "\U000024C2-\U0001F251"
        "]+",
        flags=re.UNICODE
    )
    return emoji_pattern.sub(r"", text)
```

```python
def normalize_repeated_chars(text):
    return re.sub(r"(.)\1{2,}", r"\1\1", text)
```

```python
text = "I am soooo happyyyyyyy"
print(normalize_repeated_chars(text))
```

```python
stop_words = set(stopwords.words("english"))
```

```python
def remove_stopwords(tokens):
    return [word for word in tokens if word.lower() not in stop_words]
```

```python
tokens = ["I", "love", "the", "NLP", "field"]

print(remove_stopwords(tokens))
```

```python
lemmatizer = WordNetLemmatizer()
```

```python
def lemmatize_tokens(tokens):
    return [
        lemmatizer.lemmatize(word)
        for word in tokens
    ]
```

```python
def remove_html(text):
    return text # Placeholder for actual HTML removal logic

def preprocess_tweet(text):

    # 1. Remove URLs
    text = remove_urls(text)

    # 2. Remove mentions
    text = remove_mentions(text)

    # 3. Handle hashtags
    text = handle_hashtags(text)

    # 4. Remove HTML
    text = remove_html(text)

    # 5. Lowercase
    text = lowercase(text)

    # 6. Remove emojis
    text = remove_emojis(text)

    # 7. Normalize repeated characters
    text = normalize_repeated_chars(text)

    # 8. Remove punctuation
    text = remove_punctuation(text)

    # 9. Tokenization
    tokens = word_tokenize(text)

    # 10. Remove stopwords
    tokens = remove_stopwords(tokens)

    # 11. Lemmatization
    tokens = lemmatize_tokens(tokens)

    # 12. Convert tokens back to text
    return " ".join(tokens)
```

```python
sample = df["Text"].iloc[1]

print("ORIGINAL:")
print(sample)

print("\nPROCESSED:")
print(preprocess_tweet(sample))
```

```python
df["Processed_Text"] = df["Text"].apply(preprocess_tweet)
```
